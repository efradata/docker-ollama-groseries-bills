```python
import fitz  # PyMuPDF
from PIL import Image


def ocr_with_pymupdf(file_path):
    """
    Intenta extraer texto de una imagen o PDF usando PyMuPDF.
    
    Nota:
      - PyMuPDF extrae el texto de la capa de texto de un PDF.
      - Si 'file_path' es una imagen o un PDF sin capa de texto, el resultado será vacío.
    
    :param file_path: Ruta al archivo (imagen o PDF).
    :return: Texto extraído (cadena de caracteres).
    """
    # Abrir el archivo (imagen o PDF)
    doc = fitz.open(file_path)
    # Tomar la primera página
    page = doc[0]
    # Extraer texto (esto funcionará solo si existe una capa de texto)
    text = page.get_text("text")
    doc.close()
    return text
```

```python
%%time
# Usando PyMuPDF (solo funcionará si hay capa de texto)
resultado_pymupdf = ocr_with_pymupdf('data/test1/Adobe Scan 16 de feb. de 2025 (1).pdf')

print(resultado_pymupdf)
```

```python

def ocr_with_pymupdf2(image_path):
    """
    Intenta extraer texto de una imagen utilizando PyMuPDF.
    
    Nota:
      - Este método extrae texto de la capa de texto. Si la imagen no tiene texto incrustado,
        no se realizará OCR y se retornará una cadena vacía.
      
    :param image_path: Ruta a la imagen (o PDF) a procesar.
    :return: Texto extraído (cadena de caracteres) o mensaje de error.
    """
    try:
        # PyMuPDF puede abrir tanto PDFs como algunos formatos de imagen
        doc = fitz.open(image_path)
        page = doc.load_page(0)
        text = page.get_text("text")
        doc.close()
        return text
    except Exception as e:
        return f"Error al extraer texto con PyMuPDF: {e}"
```

```python
%%time
# Usando PyMuPDF (solo funcionará si hay capa de texto)
resultado_pymupdf2 = ocr_with_pymupdf2('./data/test1/IMG_3548.JPG')

print(resultado_pymupdf2)
```

```python
from PIL import Image
import pytesseract

def ocr_with_pytesseract(image_path):
    """
    Realiza OCR sobre una imagen utilizando pytesseract.
    
    Requiere que Tesseract OCR esté instalado en el sistema.
    
    :param image_path: Ruta a la imagen.
    :return: Texto extraído por OCR.
    """
    # Abrir la imagen con Pillow
    image = Image.open(image_path)
    # Realizar OCR sobre la imagen
    text = pytesseract.image_to_string(image)
    return text
```

```python

# Usando pytesseract (realiza OCR sobre la imagen)
resultado_tesseract = ocr_with_pytesseract('data/test1/IMG_3548.JPG')
print("\nTexto extraído con pytesseract:")
print(resultado_tesseract)
```
