```python
def prompt(input_text):
    
    return [{
            'role': 'user',
            'content': input_text
        }]
```

```python
from ollama import chat, ChatResponse
```

```python
%%time
message = prompt('Write a prompt focused on llama3.2-vision model to extract the key value pairs from a bill/receipt image. The Image will be in spanish and in colombian pesos.')
response = chat(model='llama3.2-vision', messages=message)

print(response['message']['content'])
```
