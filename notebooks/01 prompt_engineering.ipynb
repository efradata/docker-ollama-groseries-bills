````python
import fitz  # PyMuPDF
from PIL import Image


def ocr_with_pymupdf(file_path):
    """
    Intenta extraer texto de una imagen o PDF usando PyMuPDF.
    
    Nota:
      - PyMuPDF extrae el texto de la capa de texto de un PDF.
      - Si 'file_path' es una imagen o un PDF sin capa de texto, el resultado será vacío.
    
    :param file_path: Ruta al archivo (imagen o PDF).
    :return: Texto extraído (cadena de caracteres).
    """
    # Abrir el archivo (imagen o PDF)
    doc = fitz.open(file_path)
    # Tomar la primera página
    page = doc[0]
    # Extraer texto (esto funcionará solo si existe una capa de texto)
    text = page.get_text("text")
    doc.close()
    return text
````

````python
from PIL import Image
import pytesseract

def ocr_with_pytesseract(image_path):
    """
    Realiza OCR sobre una imagen utilizando pytesseract.
    
    Requiere que Tesseract OCR esté instalado en el sistema.
    
    :param image_path: Ruta a la imagen.
    :return: Texto extraído por OCR.
    """
    # Abrir la imagen con Pillow
    image = Image.open(image_path)
    # Realizar OCR sobre la imagen
    text = pytesseract.image_to_string(image)
    return text


# Usando pytesseract (realiza OCR sobre la imagen)
ocr = ocr_with_pytesseract('data/test1/IMG_3548.JPG')
print("\nTexto extraído con pytesseract:")
print(ocr)
````

````python
%%time
from ollama import chat, ChatResponse

PROMPT = f"""Structure following OCR text from a Colombian grocery bill

**OCR Text**:{ocr}

**Instructions**:
1. **Store Name**: Normalize using this map:
   {{"D1 SAS": "D1", "Jeronimo Martins": "ARA", "Almacenes Éxito": "Éxito"}}

2. **Date Extraction**:
   - Convert formats like "15/08/2023" or "15-Ago-2023" to YYYY-MM-DD

3. **Item Extraction (Include ALL items)**:
   - item_name: Spanish product name
   - quantity: Default to 1 if missing
   - price_per_unit: COP value without symbols ($2.500 → 2500)

4. **Cleanup**:Extract the ONLY the items in the following OCR text
   Remove:
   - NIT/RUT numbers
   - Payment methods ("TARJETA CREDITO")
   - Total lines ("TOTAL COP $50.000")
   - Legal text ("IVA incluido")

**Example JSON Output**:
{{
  "store_name": "D1",
  "date": "2023-08-15",
  "items": [
    {{
      "item_name": "Leche Entera",
      "quantity": 2,
      "price_per_unit": 4500
    }}, ...
  ]
}}
"""

message = prompt(PROMPT)
response = chat(model='qwen2.5:7b', messages=message)

print(response['message']['content'])
````
